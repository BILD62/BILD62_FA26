### Default function values
If we usually want a function to work one way, but occasionally need it to do something else, we can allow people to pass a parameter when they need to but provide a default to make the normal case easier. The example below shows how Python matches values to parameters:

````python
def display(a=1, b=2, c=3):
    print('a:', a, 'b:', b, 'c:', c)

print('no parameters:')
display()
print('one parameter:')
display(55)
print('two parameters:')
display(55, 66)
````

As this example shows, parameters are matched up from left to right, and any that haven’t been given a value explicitly get their default value. We can override this behavior by naming the value as we pass it in:

````python
print('only setting the value of c')
display(c=77)
````

### When do you *need* `return`?

## Writing a program to count GC content

Below, there is a function to calculate the GC content of a DNA string of length 4. It includes a few new elements that we haven't discussed, but can you see what it's doing?

````python
# Write our function

def GCcontent4(DNA):
    
    '''counts GC content for a DNA string of length 4'''
    
    counter = 0 # Initialize counter

    DNA = DNA.upper() #Added this line in mine to force capitals
    
    if DNA[0]=='G' or DNA[0]=='C':
        counter = counter+1
    if DNA[1]=='G' or DNA[1]=='C':
        counter = counter+1
    if DNA[2]=='G' or DNA[2]=='C':
        counter = counter+1
    if DNA[3]=='G' or DNA[3]=='C':
        counter = counter+1
        
    return counter/4.0
````

````python
# Call our function
GCcontent4('tcga')
````

The `GCcontent4` uses **conditional statements** to test whether a given nucleotide in the sequence is equal to either a G or a C. In other words, it is doing a **value comparison**. If either of those conditions are met, it increments a **counter**. 

**Question**: Why can't we write `DNA[0] == 'C' or 'G'`?

You need to do a separate comparison of DNA[0] with 'C' and DNA[0] with 'G' to see if the first element in DNA is 'C' or if the first element in DNA is 'G'.

## Function 1: Rock, paper, scissors

In this game, one person randomly chooses rock, paper, or scissors, and the other person (you!) picks one too. Rock beats scissors, scissors beats paper, and paper beats rock; if you both pick the same thing, it's a tie.

The function below is supposed to implement rock, paper, scissors and show you if you won, lost, or tied.

**Note**: This implements a tool we haven't seen yet: `np.random.randint(1,4)`. This function picks an integer randomly from 1, 2, or 3.

**Hint**: Add a `print` statement to the function so that you can see what the computer chose!

````python
import numpy as np

def rockpaperscissors():

    user_choice = int(input('Rock (1), paper (2), or scissors (3)?'))
    computer_choice = np.random.randint(1, 4)
    print(computer_choice)
    if user_choice == computer_choice:
        print('Tie!')
    elif user_choice == 1 and computer_choice == 2:
        print('You lose!')
    elif user_choice == 2 and computer_choice == 3:
        print('You lose!')
    elif user_choice == 3 and computer_choice == 1:
        print('You lose!')
    else:
        print('You win!')
````

````python
rockpaperscissors()
````

## Function 2: Temperature Conversion

The function below should convert a temperature from Fahrenheit to Celsius and **return** that value to be used later on in the code. It is also supposed to have a docstring that you can see when you use the `help()` function.

````python
def f_to_c(fahrenheit):
    '''Converts Fahrenheit to Celsius'''
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius
````

````python
help(f_to_c)
````
