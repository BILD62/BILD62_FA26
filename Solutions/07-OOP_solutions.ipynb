## Classes

**Date** above is a class that Python has already defined for us. As it turns out, we can define classes too! Writing your own classes isn't *necessary* but it is a useful way to create complex, custom objects with their own data & methods. 

A class is defined almost like a function, but using the `class` keyword, and the class definition usually contains a number of class method definitions (a function in a class).
* Each class method should have an argument `self` as its first argument. This object is a self-reference. You'll almost never see `self` outside of a class definition.
    * Why do you need `self`? It's a little strange, but think of it this way: *an object always passes itself as the first argument to any of its own methods.*
* Some class method names have special meaning. For example, `__init__` is a method that is invoked when the object is first created.
    * (Full list <a href="https://docs.python.org/2.0/ref/specialnames.html">here</a>)
    
Let's create `Dog` as an example class.

````python
# Define a class with `class`. 
class Dog():
    
    # Class attributes for all Dogs
    sound = 'Woof'

    # Class methods for objects of type Dog
    # Self is a special parameter that refers to the object
    def speak(self):
        print(self.sound)
````

````python
# create an instance of Dog
Yeti = Dog()

# access sound attribute
print(Yeti.sound)
# use speak method
Yeti.speak()
````

````python
# Initialize a group of dogs
pack_of_dogs = [Dog(), Dog(), Dog(), Dog()]

# Write for loop so each dog can speak
for dog in pack_of_dogs:
    dog.speak()
````

We can also create **instance-specific attributes**. Below, we've added an `init method` (the code below with `__init__`) which will create the attribute `name` whenever this class is initialized. This will allow each Dog instance to have its own name. The init method is called every time we create an instance of our class.

**Note:** the order of methods and attributes does not matter, but it is conventional to have the init method first.

````python
class Dog():
    
    sound = 'Woof'
    
    # Initializer, allows us to specificy instance-specific attributes
    # leading and trailing double underscores indicates that this is special to Python
    def __init__(self,name):
        self.name = name

    def speak(self):
        print(self.sound)
````

````python
# Initialize a dog
# what goes in the parentheses is defined in the __init__
my_dog = Dog('Roger')

# Ask dog to speak
my_dog.speak()
````

````python
# Now, we need to give dog an init argument 'name'
# If not, we will receive an error
your_dog = Dog('Spot')
````

We can also define methods that take inputs. These methods are written as functions!

Below, `speak` has been modified to be proportional to the number of squirrels.

````python
class Dog():
    
    # Class attributes for all Dogs
    sound = 'Woof'
    
    # Initializer, allows us to specificy instance-specific attributes
    # leading and trailing double underscores indicates that this is special to Python
    def __init__(self,name):
        self.name = name
    
    # Squirrels is a time series of # of squirrels, which our dog sums
    def speak(self,squirrels):
        self.barks = sum(squirrels)
````

````python
# Try your dog here
squirrels = [1, 2, 3, 4]
barking_dog = Dog('Daisy')
barking_dog.speak(squirrels)
print(barking_dog.barks)
````

## Bonus: Class Inheritance
Classes can also inherit other classes! First we define a parent class, `Brain`:

````python
class Brain(): 
    
    def __init__(self, size = None, folded = None):
        self.size = size
        self.folded = folded
        
    def print_info(self):
        folded_string = ''
        if not self.folded:
            folded_string = 'not'
        print('This brain is ' + self.size + ' and is ' + folded_string + ' folded.')
        
        
my_brain = Brain(size='large',folded=True)

my_brain.print_info()
````

Now that we've created our parent class, `Brain`, we can inherit it by placing it in the parentheses in our class definition. This is telling Python that a SheepBrain is a type of Brain. By doing this, SheepBrain has access to all of the data and methods defined in Brain.

`super()` allows us to access the parent class data & methods. We need to use this when we are inheriting a class.

````python
class SheepBrain(Brain):
    
    def __init__(self, size = 'medium', folded = False):
        super().__init__(size, folded)
        
class HumanBrain(Brain):
    def __init__(self, size = 'large', folded = True):
        super().__init__(size, folded)
````

**For a more involved implementation of the Neuron class 

````python
#Below is just ONE possible implementation of a Neuron class that we can use to simulate an interacting population
#of neurons.

#Depending on your approach, your code may look pretty different!

import numpy as np
class Neuron():
    def __init__(self, diameter=10, neuron_type='pyramidal', spontaneous_firing_rate = 1):
        self.diameter = diameter
        self.neuron_type = neuron_type
        self.spontaneous_firing_rate = spontaneous_firing_rate

    def count_spikes(self, spiketrain):
        count = 0
        for is_spike in spiketrain:
            if is_spike:
                count += 1
            else:
                count = count
        self.firing_rate = count / len(spiketrain)
        return self

    def integrate(self, spike_input, leak=0.1):
        if hasattr(self, 'firing_rate'): #Checks to see if our object already has the firing_rate attribute
            self.firing_rate += leak * (self.spontaneous_firing_rate - self.firing_rate) + spike_input #Adds a leak to restore baseline firing
            self.firing_rate = max(0, self.firing_rate) #Keeps the firing rate from becoming negative
        else:
            self.firing_rate = self.spontaneous_firing_rate + spike_input
            self.firing_rate = max(0, self.firing_rate)
        return self

    def external_stim(self, stim=1, threshold=95): #This function has a random chance of producing an external stimulus
        check = np.random.randint(0,101) #Draw a random number between 0 and 100
        generate_stim = check > threshold #Creates variable that evaluates to true when check is greater than threshold
        if generate_stim:
            if hasattr(self, 'firing_rate'):
                self.firing_rate += stim
            else:
                self.firing_rate = self.spontaneous_firing_rate + stim
        return self
        
````

````python
class InhibitoryNeuron(Neuron):
    def __init__(self, diameter = 7, neuron_type='interneuron', neuron_func = 'Inhibitory'):
        super().__init__(diameter, neuron_type)
        self.neuron_func = neuron_func

class ExcitatoryNeuron(Neuron):
    def __init__(self, neuron_func = 'Excitatory'):
        super().__init__()
        self.neuron_func = neuron_func
````

````python
network = [ExcitatoryNeuron(),
           InhibitoryNeuron(),
           InhibitoryNeuron(),
           ExcitatoryNeuron(),
           InhibitoryNeuron(),
           ExcitatoryNeuron(),
           ExcitatoryNeuron(),
           InhibitoryNeuron(),
           ExcitatoryNeuron(),
           ExcitatoryNeuron()] #We assign variable network to a list containing several neurons



activity_summary = [] #We'll create an empty list to store the current spike counts
t_final = 600 #s
time = list(range(t_final+1)) #We'll create a list from 0 to t_final to keep track of time

for neuron in network:
    neuron.integrate(0) #Call integrate method for each neuron to initialize firing_rate

network_mem = [[] for cell in network] #Create an empty list to store previous activity of neighbor neurons

for t in time:
    idx = t % len(network) #Each neuron fires one at a time
    
    neighbors = [network[i] for i, neuron in enumerate(network) if i != idx] #Create list of neuron except for current neuron called neighbors
    
    neighbor_mem = network_mem[idx] #Accesses previous memory of neighbor activity
    neighbor_stim = sum([-rate if neighbors[i].neuron_func == 'Inhibitory' 
                     else rate for i, rate in enumerate(neighbor_mem)]) / (len(network) - 1) #Computes mean over neighbor activity

    neuron = network[idx]
    neuron.external_stim(stim=1,threshold=99) #Checks if there is external stimulation and integrates it
    neuron.integrate(neighbor_stim) #Integrates previous neighbor activity

    
    if network[idx].neuron_func == 'Inhibitory':
        spike_input = - (neuron.firing_rate) / (len(network) - 1)
        for neighbor in neighbors:
            neighbor.integrate(spike_input) 
    else:
        spike_input = (neuron.firing_rate) / (len(network) - 1)
        for neighbor in neighbors:
            neighbor.integrate(spike_input)

    
    spike_rate = [neun.firing_rate for neun in network] #Makes a list of firing rates over whole network
    network_activity = sum(spike_rate) #Computes spike counts over whole network at time t
    network_mem[idx] = [neib.firing_rate for neib in neighbors] #Updates memory of neighbors based on current spike rates
    activity_summary.append(network_activity) #Appends current network activity to activity_summary list
````

````python
#We can plot the activity of our simulated network using the code below

import matplotlib.pyplot as plt
plt.figure(figsize=(12, 5))
plt.bar(time, activity_summary)
plt.title("Spike Rate in Simulated Network")
plt.xlabel("Time (s)")
plt.ylabel("Spike Count")
plt.show()
````
