## Comparison and Boolean Operators

We can use comparison operators to test the relationship between two objects. These statements return Booleans.

**Booleans** are variables that store `True` or `False`. They are named after the British mathematician George Boole. He first formulated Boolean algebra, which are a set of rules for how to reason with and combine these values. This is the basis of all modern computer logic.

**Note:** Capitalization *still* matters! TRUE is not a Boolean.


| Symbol |    Operation   | Usage | Boolean Outcome |
|:------:|:--------------:|:-----:|:-------:|
|    ==   |  is equal to  |`10==5*2`| True | 
|    !=   | is not equal to | `10!=5*2` | False |
|    >  | Greater than |  `10 > 2` | True |
|    <   |    Less than    |  `10 < 2` | False |
| >= | Greater than _or_ equal to | `10 >= 10` | True |
| <= | Less than _or_ equal to | `10 <= 10` | True |


<div class="alert alert-success"><b>Task:</b> Test each of the comparison operators, saving their output to a variable.</div>

```python
# Test the comparison operators here
is_equal = 7 == 3 % 6
is_not_equal = 4 != 20 // 5
is_greater_or_equal = 9 // 4 >= 2
is_greater = 7 * 3 > 6 * 4
is_less = 8 < 10
is_less_or_equal = 1 <= 1
print(is_equal, is_not_equal, is_greater_or_equal, is_greater, is_less, is_less_or_equal)
```

**Boolean operators** use Boolean logic, and include:
- `and` : True if both are true
- `or` : True if at least one is true
- `not` : Reverses the Boolean

## Conditionals
**Conditionals** are statements that check for a condition, using the `if` statement, and then only execute a set of code if the condition evaluates as `True`.

- `if`
- `elif` (else if): After an if, you can use elif statements to check additional conditions.
- `else`: After an if, you can use an else that will run if the conditional(s) above have not run.

### If/elif/else syntax
- Indentation matters! Your statements in the `if` block need to be indented by a tab or four spaces.
- You need a colon after `if`, `elif`, and `else`

```python
condition = False

if condition:
    print('This code executes if the condition evaluates as True.')
    
else: 
    print('This code executes if the condition evaluates as False')
```

### Properties of Conditionals
- Conditionals can take any expression that can be evaluated as `True` or `False`. 
- The order of conditional blocks is always `if` then `elif`(s) then `else`.
- `elif` must come before `else`.
- An `else` statement is not required, but if both the `if` and the `elif` conditions are not met (both evaluate as `False`), nothing will occur
- **At most one component (`if` / `elif` / `else`) of a conditional will run**

```python
condition_1 = False
condition_2 = True

if condition_1:
    print('This code executes if condition_1 evaluates as True.')
elif condition_2:
    print('This code executes if condition_2 evaluates as True AND condition_1 evaluates as False.')  
else: 
    print('This code executes if both condition_1 and condition_2 evaluate as False')
```

Remember that all of the comparison operators above ultimately are evaluated as `True` or `False`. That means statements using them can also be used in our conditionals:

```python
if 6 < 10 and 4 == 4:
    print('This code executes if the condition evaluates as True.')
```

## Putting it together: Booleans in a conditional

A comparison produces a Boolean, and a conditional uses a Boolean to decide which block of code to run. Here, we store the result of a comparison in a variable and then use that variable in an `if` statement.

### Code tracing
**Code tracing** means stepping through code one line at a time, *by hand*, and keeping track of what each variable holds and what the program does at each step. It's one of the best ways to understand what code does (and to find bugs) before you run it.

```python
cell_count = 12                   #Assigns the int 12 to the variable cell_count
is_large = cell_count > 10        #Compares the value stored in cell_count agains the int 10 and assigns the resulting boolean to the variable is_large

if is_large:                      #Checks if is_large evaluates to True
    print('large colony')         #If is_large is True, then display large colony
else:                             #Runs when is_large evaluates to False
    print('small colony')         #If is_large is False, then display small colony
```
