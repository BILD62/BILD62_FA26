# Working with Lists, Tuples, and Dictionaries

As we continue our Python journey, we'll frequently encounter new types of objects and learn how to work with them. This week, we're focused on objects that can store information called **data structures**.

Several types of data structures exist including **lists**, **tuples**, and **dictionaries**. Each of these is summarized in the table below:

| Datastructure Type |Constructor| Description | Important Methods|
|--------------------|-----------|-------------|------------------|
|Lists               |[]         |Mutable data structure that contains ordered data| .append() - adds item to list|
|Tuples              |()         |Immutable data structure that contains ordered data| - |
|Dictionaries        |{}         |Flexible data structure that uses key-value pairs instead of numerical indexing| .update({'key' : value}) - adds key-value pair to dictionary |
